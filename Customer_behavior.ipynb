```python
import pandas as pd

df=pd.read_csv('D:\customer_shopping_behavior.csv')
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x : x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
bins=[0,18,30,50,100]
labels=['Young Adult','Adult','Middle Aged','Senior']
df['age_group']=pd.cut(df['age'],bins=bins,labels=labels)
```

```python
df[['age','age_group']].head(10)
```

```python
df['frequency_of_purchases'].unique()
```

```python
frequancy={
    'Fortnightly' : 14,
    'Weekly' : 7,
    'Annually' :365,
    'Quarterly': 90,
    'Bi-Weekly' :14,
    'Monthly' :30,
    'Every 3 Months': 90
}
df['purchases_frequancy_days']=df['frequency_of_purchases'].map(frequancy)
```

```python
df[['purchases_frequancy_days','frequency_of_purchases']]
```
